```python
from ultralytics import YOLO

DATA_YAML = 'src/data.yaml'
WEIGHTS = 'yolo11s.pt'
RUN_NAME = 'shahed_v2'

model = YOLO(WEIGHTS)

model.train(
    data=DATA_YAML,
    epochs=100,
    imgsz=800,
    batch=16,
    workers=2,
    device=0,
    name=RUN_NAME,
    cache=False,
    patience=0
)
```

# Shahed Interceptor - Training

Train YOLO11 on custom Shahed drone dataset.

```python
from ultralytics import YOLO
```

## Setup
Before running, make sure you have:
- Your dataset with `data.yaml` in the right path
- A pretrained checkpoint (or use yolov8n.pt for fresh training)

```python
# paths - change these to match your setup
DATA_YAML = 'path/to/your/data.yaml'  # dataset config
WEIGHTS   = 'path/to/last.pt'         # pretrained weights (or yolov8n.pt)
RUN_NAME  = 'shahed_v1'               # output folder name

model = YOLO(WEIGHTS)

model.train(
    data=DATA_YAML,
    epochs=50,
    imgsz=928,
    batch=8,
    workers=1,
    device=0,
    name=RUN_NAME
)
```

## Results
After training, check `runs/detect/<RUN_NAME>/` for:
- `weights/best.pt` - best model
- `results.csv` - training metrics
- `confusion_matrix.png`
